```python
# Operators and Expressions on variables
x=20
y=10

# integer operations
print(f"the value of x+y is {x+y}")
print(f"the value of x-y is {x-y}")
print(f"the value of x*y is {x*y}")
print(f"the value of x/y is {x/y}")
print(f"the value of x//y is {x//y}") # floor division operator returns the largest integer less than or equal to the division of the two operands.  
print(f"the value of x%y is {x%y}") # modulus operator returns the remainder of the division of the two operands.
print(f"the value of x**y is {x**y}") # exponentiation operator returns the value of x raised to the power of y.
print(f"the value of x**y is {pow(x,y)}") #pow is a function to calculate power of a number
print(f"the value of x**y is {x**y}") #** is a operator to calculate power of a number
```

```python
# palindrome check using floor division operator for integer division and modulus operator for remainder
def is_palindrome(n):
    original = n
    reversed_num = 0

    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n //= 10

    return original == reversed_num
```

```python
# Arithmetic operators
print("Arithmetic operators")
print(f"x + y = {x + y}")
print(f"x - y = {x - y}")
print(f"x * y = {x * y}")
print(f"x / y = {x / y}")
print(f"x // y = {x // y}")
print(f"x % y = {x % y}")
print(f"x ** y = {x ** y}")

# Comparison operators
print("\nComparison operators")
print(f"t > y -> {t > y}")
print(f"r >= 10.0 -> {r >= 10.0}")
print(f"s == 'hi' -> {s == 'hi'}")
print(f"NAME.strip() == 'Siva varaprasad' -> {NAME.strip() == 'Siva varaprasad'}")

# Logical operators
print("\nLogical operators")
print(f"(x > y) and (y > 0) -> {(x > y) and (y > 0)}")
print(f"(x < y) or (y > 0) -> {(x < y) or (y > 0)}")
print(f"not k -> {not k}")
print(f"w is None -> {w is None}")

# String operators
print("\nString operators")
print(f"s + ' ' + NAME.strip() -> {s + ' ' + NAME.strip()}")
print(f"(s + ' ') * 3 -> {(s + ' ') * 3}")
print(f"len(s) = {len(s)}")
print(f"'hi' in s -> {'hi' in s}")
print(f"'siva' in NAME.lower() -> {'siva' in NAME.lower()}")

# Sequence operators (list and tuple)
print("\nSequence operators")
print(f"z + [5, 6] -> {z + [5, 6]}")
print(f"z[0] = {z[0]}, z[-1] = {z[-1]}")
print(f"ss[0] = {ss[0]}, ss[-1] = {ss[-1]}")
print(f"3 in z -> {3 in z}")
print(f"5 in ss -> {5 in ss}")
print(f"len(z) = {len(z)}, len(ss) = {len(ss)}")

# Dictionary operators
print("\nDictionary operators")
print(f"dt['name'] = {dt['name']}")
print(f"'age' in dt -> {'age' in dt}")
print(f"dt.get('age') = {dt.get('age')}")
print(f"len(dt) = {len(dt)}")

# Identity and membership operators
print("\nIdentity and membership operators")
print(f"s is 'hi' -> {s is 'hi'}")
print(f"z is not ss -> {z is not ss}")
print(f"1 in z -> {1 in z}")
print(f"4 not in z -> {4 not in z}")

# Assignment expressions / compound operators
print("\nAssignment and compound ops")
a = 10
a += 5
print(f"a after a += 5: {a}")
a *= 2
print(f"a after a *= 2: {a}")
a %= 7
print(f"a after a %= 7: {a}")

# Date/time operators
print("\nDate/time examples")
print(f"date_variable + datetime.timedelta(days=7) = {date_variable + datetime.timedelta(days=7)}")
print(f"time_variable.strftime('%Y-%m-%d %H:%M:%S') = {time_variable.strftime('%Y-%m-%d %H:%M:%S')}")

# Boolean expressions
print("\nBoolean expressions")
print(f"bool(t) -> {bool(t)}")
print(f"bool(r) -> {bool(r)}")
print(f"bool(w) -> {bool(w)}")
print(f"bool(z) -> {bool(z)}")
```
