```python
def divide_numbers(a, b):
  try:
    result = a / b
  # except Exception as err : #
  #   print(f"Error: Cannot divide by zero and the error is {err}")
  except Exception as err1:
    print(f"Error: Both inputs must be numbers. {err1}")
  else:
    print(f"Success! The result is {result}.")
  finally:
    print("Execution complete.")


# divide_numbers(10, 2)
# # Output:
# # Success! The result is 5.0.
# # Execution complete.

# divide_numbers(10, 0)

divide_numbers('rama',1)
# Output:
# Error: Cannot divide by zero.
# Execution complete.
```

```python
import math
import os
import sys
import weakref

def run_demo(title, func):
    try:
        func()
    except BaseException as e:
        print(f"[{type(e).__name__}] {title} -> {e}")

print("--- 1. SYNTAX & INDENTATION ERRORS (Evaluated dynamically) ---")
run_demo("SyntaxError", lambda: eval("x === 1"))
run_demo("IndentationError", lambda: exec("if True:\nprint('No indent')"))
run_demo("TabError", lambda: exec("if True:\n\tprint('tab')\n    print('spaces')"))

print("\n--- 2. DATA, TYPE & LOOKUP ERRORS ---")
run_demo("TypeError", lambda: "Age: " + 25)
run_demo("ValueError", lambda: int("not_a_number"))
run_demo("NameError", lambda: eval("undefined_variable"))

def unbound_local_example():
    print(x)
    x = 10
run_demo("UnboundLocalError", unbound_local_example)

run_demo("AttributeError", lambda: [1, 2, 3].non_existent_method())
run_demo("IndexError", lambda: [10, 20][5])
run_demo("KeyError", lambda: {"a": 1}["missing_key"])

print("\n--- 3. ARITHMETIC ERRORS ---")
run_demo("ZeroDivisionError", lambda: 10 / 0)
run_demo("OverflowError", lambda: math.exp(1000))

print("\n--- 4. IMPORT ERRORS ---")
run_demo("ModuleNotFoundError", lambda: __import__("non_existent_module_xyz"))
def import_err_example():
    from math import non_existent_function
run_demo("ImportError", import_err_example)

print("\n--- 5. FILE & OS ERRORS ---")
run_demo("FileNotFoundError", lambda: open("non_existent_file_12345.txt"))
run_demo("IsADirectoryError", lambda: open("."))
run_demo("NotADirectoryError", lambda: os.listdir(__file__))

def file_exists_example():
    os.makedirs("temp_test_dir_demo", exist_ok=True)
    os.mkdir("temp_test_dir_demo") # Fails because it already exists
try:
    run_demo("FileExistsError", file_exists_example)
finally:
    if os.path.exists("temp_test_dir_demo"):
        os.rmdir("temp_test_dir_demo")

print("\n--- 6. ITERATION & GENERATOR ERRORS ---")
def stop_iter_example():
    it = iter([1])
    next(it)
    next(it) # No items left
run_demo("StopIteration", stop_iter_example)

def gen_exit_example():
    def my_gen():
        try:
            yield 1
        except GeneratorExit as e:
            print(f"[GeneratorExit] Triggered when generator .close() was called")
            raise
    g = my_gen()
    next(g)
    g.close()
gen_exit_example()

print("\n--- 7. SYSTEM, CONTROL FLOW & MEMORY ERRORS ---")
def assert_example():
    assert 2 + 2 == 5, "Math logic failed!"
run_demo("AssertionError", assert_example)

def recursion_example():
    return recursion_example()
run_demo("RecursionError", recursion_example)

def not_implemented_example():
    raise NotImplementedError("Subclasses must override this method")
run_demo("NotImplementedError", not_implemented_example)

def runtime_err_example():
    d = {"a": 1, "b": 2}
    for k in d:
        d["c"] = 3 # Modifying dict size during iteration
run_demo("RuntimeError", runtime_err_example)

def reference_err_example():
    class TempObj: pass
    obj = TempObj()
    proxy = weakref.proxy(obj)
    del obj # Garbage collected
    print(proxy.attr)
run_demo("ReferenceError", reference_err_example)

def buffer_err_example():
    b = bytearray(b"hello")
    view = memoryview(b)
    b.append(1) # Cannot resize while memoryview is active
run_demo("BufferError", buffer_err_example)

run_demo("SystemExit", lambda: sys.exit("Exited safely inside try/except"))

print("\n--- 8. UNICODE & ENCODING ERRORS ---")
run_demo("UnicodeEncodeError", lambda: "🐍".encode("ascii"))
run_demo("UnicodeDecodeError", lambda: b"\xff".decode("utf-8"))

print("\n--- 9. SPECIAL INTERACTIVE / SYSTEM ERRORS (Commented out) ---")
# 1. KeyboardInterrupt: Press Ctrl + C while running a loop
# 2. EOFError: Run input() and press Ctrl + D (Mac/Linux) or Ctrl + Z (Windows)
# 3. MemoryError: Run `[0] * (10**12)` (Warning: may freeze your system!)
# 4. PermissionError: Try opening a protected system file like `open('/etc/shadow', 'w')`
```
